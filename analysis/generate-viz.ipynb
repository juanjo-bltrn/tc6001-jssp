```python
from pathlib import Path
from analysis.visualizer import plot_jssp_schedule
```

```python
def generate_viz(ruta_directorio):
    directorio = Path(ruta_directorio)
    
    # Recorre de forma recursiva todos los archivos en el directorio y subdirectorios
    for archivo in directorio.rglob('*'):
        print(archivo)
        # Verifica que sea un archivo y tenga extensión .txt (sin importar mayúsculas/minúsculas)
        if archivo.is_file() and archivo.suffix.lower() == '.txt':
            path_viz = str(archivo.with_name('schedule-' + archivo.name.split(".")[0] + ".png"))
            print(f"Ploting schedule for {archivo}")
            plot_jssp_schedule(file_path=str(archivo),
                               save_path=path_viz,
                               show=False)
```

```python
# Reemplaza la ruta por el directorio que deseas procesar
generate_viz(r"solutions")
```
